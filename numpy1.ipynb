```python
import numpy as np
import time
```

```python
size = 10_000_0
py_list = list(range(size))
start = time.time()
sq_list = (x**2 for x in py_list)
end = time.time()
print(f"python list time = {end - start}" )

np_arr = np.array(py_list)
start = time.time()
sq_arr = np_arr ** 2
end = time.time()
print(f"np_array time = {end - start}")
```
